# Corner Deviation Analysis

In this notebook, we will analyze the corner deviation data, comparing floor and ceiling deviations, and visualize the results.

```python
import sys
sys.path.append('.')

from src.plot_utils import PlotUtils
import matplotlib.pyplot as plt
%matplotlib inline
```

## Creating Plots

First, we'll call the `draw_plots` method to create the plots and get the statistics.

```python
plot_paths = PlotUtils.draw_plots()
print(f"Created plots: {plot_paths}")
```

## Visualizing Plots

Now, let's display the created plots in the notebook.

```python
for path in plot_paths:
    plt.figure(figsize=(12, 8))
    img = plt.imread(path)
    plt.imshow(img)
    plt.axis('off')
    plt.title(path)
    plt.show()
```

## Analyzing Results

Based on the generated plots and output statistics, we can draw the following conclusions:

1. Model Accuracy:
   - The comparison plot of ground truth vs. predicted corners shows how accurately the model determines the number of corners in rooms.
   - We can see the correlation between predicted and actual corner counts.

2. Deviation Distribution:
   - The box plot of the deviations gives insight into the distribution of the deviations (mean, max and min).
   - This helps identify potential outliers and understand the spread of deviations.

3. Mean Deviation Patterns:
   - The histogram of mean deviations allows us to assess which deviation values occur most frequently.
   - This helps understand the typical error ranges in the model's predictions.

4. Floor vs Ceiling Analysis:
   - The floor vs ceiling comparison plot shows how deviations differ between floor and ceiling measurements.
   - This helps identify if the model performs better on floor or ceiling corner detection.

5. Room-specific Analysis:
   - Different room types show varying levels of accuracy and deviation.
   - Some room types may be more challenging for the model than others.
